```python
class Atm:

    __counter = 0

    def __init__(self):
        self.__pin = ""
        self.__balance = 0

        Atm.__counter += 1
        print("\nWelcome to Python Banking System")
        print("ATM ID:", id(self))

        self.__menu()

    @staticmethod
    def get_counter():
        return Atm.__counter

    def __menu(self):
        while True:
            print("\n" + "="*35)
            print("        MAIN MENU")
            print("="*35)
            print("1. Create PIN")
            print("2. Deposit Money")
            print("3. Withdraw Money")
            print("4. Check Balance")
            print("5. Exit")
            print("="*35)

            user_input = input("Select an option (1-5): ")

            if user_input == "1":
                self.create_pin()
            elif user_input == "2":
                self.deposit()
            elif user_input == "3":
                self.withdraw()
            elif user_input == "4":
                self.check_balance()
            elif user_input == "5":
                print("\n Thank you for using Python Banking System!")
                break
            else:
                print(" Invalid selection. Please try again.\n")

    def create_pin(self):
        while True:
            print("\n PIN Setup")
            new_pin = input("Enter a 4-digit PIN: ")

            if new_pin.isdigit() and len(new_pin) == 4:
                self.__pin = new_pin
                print(" PIN created successfully!\n")
                break
            else:
                print(" Invalid PIN! Must be exactly 4 digits.\n")

    def __verify_pin(self):
        print("\n Authentication Required")
        temp = input("Enter your PIN: ")
        return temp == self.__pin

    def deposit(self):
        if self.__verify_pin():
            try:
                print("\n Deposit Transaction")
                amount = int(input("Enter amount to deposit: ₹"))

                if amount > 0:
                    self.__balance += amount
                    print(f"₹{amount} deposited successfully.")
                    print(f"Available Balance: ₹{self.__balance}\n")
                else:
                    print("Please enter a valid amount.\n")
            except:
                print("Invalid input. Please enter numbers only.\n")
        else:
            print("Incorrect PIN. Transaction denied.\n")

    def withdraw(self):
        if self.__verify_pin():
            try:
                print("\n Withdrawal Transaction")
                amount = int(input("Enter amount to withdraw: ₹"))

                if amount <= self.__balance and amount > 0:
                    self.__balance -= amount
                    print(f"₹{amount} withdrawn successfully.")
                    print(f"Remaining Balance: ₹{self.__balance}\n")
                elif amount > self.__balance:
                    print("Insufficient balance.\n")
                else:
                    print("Enter a valid amount.\n")
            except:
                print("Invalid input. Please enter numbers only.\n")
        else:
            print("Incorrect PIN. Transaction denied.\n")

    def check_balance(self):
        if self.__verify_pin():
            print("\nAccount Summary")
            print("-"*30)
            print(f"Available Balance: ₹{self.__balance}")
            print("-"*30 + "\n")
        else:
            print("Incorrect PIN. Access denied.\n")
```
